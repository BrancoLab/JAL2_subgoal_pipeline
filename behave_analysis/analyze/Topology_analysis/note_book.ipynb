```python
import numpy as np
from ripser import ripser
from persim import plot_diagrams
import numpy as np
from sklearn import decomposition
import ripserplusplus as rpp_py
```

# Rapids library

```python
pca = decomposition.PCA(n_components=15)
pca.fit(matrix)
transformed_data = pca.transform(matrix)
```
