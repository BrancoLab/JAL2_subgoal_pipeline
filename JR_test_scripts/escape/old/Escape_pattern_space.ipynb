```python
def extract_escape_time(session, frame_by_cluster_matrix, behave, y_pos, x_pos, bar, barflip, compression_var):
    # extract the time around escapes
    start = [0]
    for tr, of in enumerate(session.audio.onset_frames):
        if isinstance(of, list): of = of[0]
        if isinstance(of, np.ndarray): of = of[0]
        # find actual length of time until mouse is in shelter (or 5s if he never makes it)
        y_loc = y_pos[of:of + (20*40)]
        x_loc = x_pos[of:of + (20*40)]
        in_shelt_y = y_loc > session.shelter_location[0][1]
        in_shelt_x = np.logical_and(x_loc > session.shelter_location[0][0],x_loc < session.shelter_location[1][0])
        in_shelt = np.where(np.logical_and(in_shelt_x, in_shelt_y))[0]
        if len(in_shelt) == 0: in_shelt = 5*40 # cases when mouse never reaches shelter
        else: in_shelt = in_shelt[0]+40

        # extract variables
        neur = frame_by_cluster_matrix[of:of + in_shelt,:] # time x neurons
        this_speed = behave[of:of + in_shelt]
        this_y = y_pos[of:of + in_shelt]
        this_x = x_pos[of:of + in_shelt]

        # condition vector
        c = np.zeros((len(y_pos[of:of + in_shelt])))
        if bar[of] == True: c += 1
        if barflip[of] == True: c += 1

        if compression_var == 'distance_shelter':
            var = compute_dist_shelt(this_x, this_y, c, session)
        elif compression_var == 'y_pos':
            var = this_y
        elif compression_var == 'escape':
            dd = compute_escape_trajectory(this_x, this_y)
            var = (dd/np.amax(dd))*1000 # fraction of total distance, *1000 for discretizing in bins of 10

        disc_var = discretize_x_axis(var)
        new_activity, new_pos = compress_vars(disc_var, neur.T) # new_activity is neurons x time

        # concatenate trials
        if tr == 0:
            pre_mean = np.mean(frame_by_cluster_matrix[of - 40:of,:], axis = 0)
            post_mean = np.mean(frame_by_cluster_matrix[of:of+(5*40),:], axis = 0)
            escape_matrix = new_activity
            esc_var = new_pos
        else:
            escape_matrix = np.hstack((escape_matrix,new_activity))
            esc_var = np.append(esc_var, new_pos)
        
        start.append(len(esc_var))

    # z-score and keep only relevant neurons (max. 200)
    spks = zscore(escape_matrix, axis=1) # need neurons x time
    interesting_neurons = pre_mean < post_mean
    fit_spks = spks[interesting_neurons,:]
    if np.shape(fit_spks)[0] > 200:
        sorter = np.argsort(post_mean[interesting_neurons] - pre_mean[interesting_neurons])
        fit_spks = fit_spks[sorter,:]
        fit_spks = fit_spks[-200:,:]
    return esc_var, fit_spks, start[:-1]
```

```python
def run_rastermap(fit_spks, fit_type = 'single_neurons'):
  # fit_spks is of shape neurons x time
  # rastermap with running speed and y position
  # fit_spks = fit_spks[:,:6*40]

  if fit_type == 'single_neurons':
    model = Rastermap(n_clusters=None, # None turns off clustering and sorts single neurons 
                      n_PCs=64, # use fewer PCs than neurons
                      locality=0.15, # some locality in sorting (this is a value from 0-1)
                      time_lag_window=40, # use future timepoints to compute correlation
                      grid_upsample=0, # 0 turns off upsampling since we're using single neurons
                    ).fit(fit_spks)
  elif fit_type == 'clustering':
    n_neurons = np.shape(fit_spks)[0]
    model = Rastermap(n_clusters=100,#int(np.round(n_neurons/20)), # number of clusters to compute
                  n_PCs=64, # number of PCs to use
                  locality=0., # locality in sorting to find sequences (this is a value from 0-1)
                  grid_upsample=4, # default value, 10 is good for large recordings
                ).fit(fit_spks)
  # y = model.embedding # neurons x 1
  isort = model.isort
  return isort
```

```python
def activity_sorting(frame_by_cluster_matrix, trials):
    """sort the matrix based on the average time-based activity on a few trials"""
    c = 0
    sub_matrix = np.empty((6*40,np.shape(frame_by_cluster_matrix)[1],len(trials)))
    for tr, of in enumerate(session.audio.onset_frames):
        if isinstance(of, list): of = of[0]
        if isinstance(of, np.ndarray): of = of[0]
        if tr == 0:
            pre_mean = np.mean(frame_by_cluster_matrix[of - 40:of,:], axis = 0)
            post_mean = np.mean(frame_by_cluster_matrix[of:of+(5*40),:], axis = 0)
        if tr in trials:
            in_shelt = 5*40 # cases when mouse never reaches shelter
            sub_matrix[:,:,c] = frame_by_cluster_matrix[of - 40:of + in_shelt,:]
            c += 1

    interesting_neurons = pre_mean < post_mean
    fit_spks = sub_matrix[:,interesting_neurons,:]
    if np.shape(fit_spks)[1] > 200:
        sorter = np.argsort(post_mean[interesting_neurons] - pre_mean[interesting_neurons])
        fit_spks = fit_spks[:,sorter,:]
        fit_spks = fit_spks[:,-200:,:]
    # average in bins
    angles_firing = np.mean(fit_spks, axis = 2)
    idx = np.argmax(angles_firing, axis = 0)
    isort = np.argsort(idx)
    return isort   
```

```python
def plot_rastermap(fit_spks, esc_var, session, isort, nickname, compression_var, start = []):
    # make figure with grid for easy plotting
    fig = plt.figure(figsize=(12,6), dpi=200)
    grid = plt.GridSpec(9, 20, figure=fig, wspace = 0.05, hspace = 0.3)

    # plot running speed
    ax = plt.subplot(grid[0, :-1])
    ax.plot(esc_var, color='r')
    ax.set_xlim((0,len(esc_var)))
    ax.axis("off")

    # plot superneuron activity
    ax = plt.subplot(grid[1:, :-1])
    ax.imshow(fit_spks[isort,:], cmap="gray_r", vmin=0, vmax=1.2, aspect="auto", interpolation = "none")
    if len(start) == 0:
        start = 40
        for i in np.arange(len(session.audio.onset_frames)):
            ax.plot([start,start],[0,len(isort)],'--b')
            start += 6*40
    else:
        for i in start:
            ax.plot([i,i],[0,len(isort)],'--b')
    ax.set_xlabel(compression_var)
    ax.set_ylabel("neurons")

    stim_resp_path = make_directory(os.path.join(session.base_path, session.processed_path, "stim_resp", "rastermap"))
    save_path=str(stim_resp_path) + "/" + nickname + ".png"
    fig.savefig(save_path)

    dump_path = "Z:\Jasmine_Laurence\escape_rastermap"
    fig.savefig(dump_path + "/" + nickname + ".png")
    plt.close()
```
