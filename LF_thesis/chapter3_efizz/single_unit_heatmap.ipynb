```python
# Single-unit heatmaps (align on video 'frames' and spikes 'spike_aligned_to_frame')

import os
import numpy as np
import polars as pl
import pandas as pd
import matplotlib.pyplot as plt

# ---- User paths ----
PROCESSED_DIR = r"Z:\Jasmine_Laurence\Experimental_Data\JAL006\JAL006_shelter_barrier_flip_5_2024_03_25T11_05_33\processed_data"
SAVE_DIR = r"Z:\Laurence\thesis\efizz_chapter\single_unit_heatmaps"
FPS = 40
NBINS = 30
COLORMAP = "magma"
DPI = 300
os.makedirs(SAVE_DIR, exist_ok=True)

# ---- Helpers ----
def add_bins_pd(df_pd, nbins=NBINS, x_col="x", y_col="y"):
    x_min, x_max = np.nanmin(df_pd[x_col].values), np.nanmax(df_pd[x_col].values)
    y_min, y_max = np.nanmin(df_pd[y_col].values), np.nanmax(df_pd[y_col].values)
    eps = 1e-9
    x_edges = np.linspace(x_min, x_max + eps, nbins + 1)
    y_edges = np.linspace(y_min, y_max + eps, nbins + 1)
    df_pd["x_bins"] = np.clip(np.digitize(df_pd[x_col].values, x_edges) - 1, 0, nbins - 1)
    df_pd["y_bins"] = np.clip(np.digitize(df_pd[y_col].values, y_edges) - 1, 0, nbins - 1)
    return x_edges, y_edges

def robust_min_max(arr, low=2, high=98):
    finite_vals = arr[np.isfinite(arr)]
    if finite_vals.size == 0:
        return 0.0, 1.0
    vmin = np.percentile(finite_vals, low)
    vmax = np.percentile(finite_vals, high)
    if np.isclose(vmin, vmax):
        vmax = vmin + 1e-6
    return float(vmin), float(vmax)

def spikes_per_sec_grid(df_cond, df_cond_clu, nbins=NBINS, fps=FPS, spike_col="spike_count"):
    # Occupancy (frames per bin) across all frames in this condition
    occ = df_cond.groupby(["y_bins", "x_bins"]).size().unstack(fill_value=0)
    # Spike counts per bin for this unit
    if df_cond_clu.empty:
        spk = pd.DataFrame(0, index=occ.index, columns=occ.columns)
    else:
        spk = df_cond_clu.groupby(["y_bins", "x_bins"])[spike_col].sum().unstack(fill_value=0)
        spk = spk.reindex(index=occ.index, columns=occ.columns, fill_value=0)
    with np.errstate(divide="ignore", invalid="ignore"):
        rate = spk.values / np.where(occ.values == 0, np.nan, occ.values) * fps
    return rate

def add_colorbar(fig, im, cax):
    cbar = fig.colorbar(im, cax=cax)
    cbar.set_label("Spikes/s", rotation=90)

# ---- Load ----
video_csv = os.path.join(PROCESSED_DIR, "full_video_dataframe.csv")
spike_csv = os.path.join(PROCESSED_DIR, "spike_count_by_frame_and_goodcluster.csv")
vdf_raw = pl.read_csv(video_csv)   # columns include: 'frames','mouse_x_position','mouse_y_position',...
sdf_raw = pl.read_csv(spike_csv)   # columns: 'spike_aligned_to_frame','spike_clusters','spike_count'

# Align columns explicitly
if "frames" not in vdf_raw.columns:
    raise ValueError("Expected 'frames' in video dataframe.")
if "spike_aligned_to_frame" not in sdf_raw.columns:
    raise ValueError("Expected 'spike_aligned_to_frame' in spike dataframe.")

vdf = vdf_raw.rename({
    "frames": "frame",
    "mouse_x_position": "x",
    "mouse_y_position": "y"
}).with_columns(pl.col("frame").cast(pl.Int64))

sdf = sdf_raw.rename({
    "spike_aligned_to_frame": "frame",
    "spike_count": "spike_count",
    "spike_clusters": "spike_clusters"
}).with_columns(pl.col("frame").cast(pl.Int64))

# Keep minimal video columns; add 'all_time' condition (no explicit condition column present)
vdf_small = vdf.select(["frame", "x", "y"])
merged = vdf_small.join(sdf.select(["frame","spike_count","spike_clusters"]), on="frame", how="left")
merged = merged.with_columns(pl.lit("all_time").alias("condition"))
merged = merged.drop_nulls(["x","y"])
merged = merged.with_columns(pl.col("spike_count").fill_null(0))

# To pandas for pivoting
df = merged.to_pandas()

# Bin positions
x_edges, y_edges = add_bins_pd(df, nbins=NBINS, x_col="x", y_col="y")

# Conditions and units
cond_values = ["all_time"]  # only all_time since no per-condition column in video CSV
unit_ids = np.sort(df["spike_clusters"].dropna().unique().astype(int))

# ---- Plot per unit ----
for clu in unit_ids:
    # Color scale from all_time
    df_all = df.copy()
    rate_all = spikes_per_sec_grid(df_all, df_all[df_all["spike_clusters"] == clu], nbins=NBINS, fps=FPS)
    vmin, vmax = robust_min_max(rate_all, 2, 98)

    fig, ax = plt.subplots(1, 1, figsize=(5, 4.5))
    cbar_ax = fig.add_axes([0.90, 0.2, 0.03, 0.6])

    df_cond = df  # all_time
    df_clu = df_cond[df_cond["spike_clusters"] == clu]
    rate = spikes_per_sec_grid(df_cond, df_clu, nbins=NBINS, fps=FPS)

    rate_masked = np.ma.array(rate, mask=~np.isfinite(rate))
    im = ax.imshow(
        rate_masked, origin="lower", cmap=COLORMAP,
        vmin=vmin, vmax=vmax, interpolation="nearest",
        extent=[x_edges[0], x_edges[-1], y_edges[0], y_edges[-1]],
        aspect="equal"
    )
    ax.set_title("all time", fontsize=14)
    ax.set_xticks([]); ax.set_yticks([])
    for s in ["top","right","left","bottom"]:
        ax.spines[s].set_visible(False)

    add_colorbar(fig, im, cbar_ax)
    total_spikes = int(df_clu["spike_count"].sum())
    fig.suptitle(f"Unit {clu} • firing rate (spikes/s) • total spikes={total_spikes}", fontsize=16)
    fig.tight_layout(rect=[0.03, 0.03, 0.88, 0.97])

    out_png = os.path.join(SAVE_DIR, f"unit_{clu}_heatmap.png")
    # fig.savefig(out_png, dpi=DPI, bbox_inches="tight")
    # # fig.savefig(os.path.join(SAVE_DIR, f"unit_{clu}_heatmap.eps"), format="eps", dpi=DPI, bbox_inches="tight")
    plt.show()
    plt.close(fig)

print(f"Saved unit heatmaps to: {SAVE_DIR}")
```
